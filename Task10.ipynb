```python
import numpy as np
from sklearn.datasets import load_iris

print("Libraries imported successfully!")
```

```python
iris = load_iris()

X = iris.data
y = iris.target

print("Iris dataset loaded successfully!")
print("Shape of dataset:", X.shape)
```

```python
print("First 10 rows of Iris Dataset:")
print(X[:10])
```

```python
print("Feature Names:")
print(iris.feature_names)
```

```python
#Positive Indexing
print("First row:")
print(X[0])

print("\nSecond row:")
print(X[1])

print("\nThird row:")
print(X[2])
```

```python
#Access Individual Elements
print("First row, first feature:", X[0, 0])
print("First row, second feature:", X[0, 1])
print("Second row, third feature:", X[1, 2])
print("Third row, fourth feature:", X[2, 3])
```

```python
#Negative Indexing
print("Last row:")
print(X[-1])

print("\nSecond last row:")
print(X[-2])

print("\nLast value of last row:")
print(X[-1, -1])
```

```python
#Slicing Rows
print(X[:5])
```

```python
print(X[10:15])
```

```python
print(X[-5:])
```

```python
#Slicing Columns
sepal_length = X[:, 0]

print("Sepal Length:")
print(sepal_length)
```

```python
sepal_width = X[:, 1]

print("Sepal Width:")
print(sepal_width)
```

```python
petal_length = X[:, 2]

print("Petal Length:")
print(petal_length)
```

```python
petal_width = X[:, 3]

print("Petal Width:")
print(petal_width)
```

```python
#Multiple Column Slicing
print("First two columns:")
print(X[:, :2])
```

```python
print("Last two columns:")
print(X[:, 2:])
```

```python
#Combined Row and Column Slicing
print("First 5 rows and first 2 columns:")
print(X[:5, :2])
```

```python
print("Last 10 rows in reverse order:")
print(X[-1:-11:-1])
```

```python
print("Every second row from the first 10 rows:")
print(X[:10:2])
```

```python
first_10 = X[:10]

print("First 10 Iris samples:")
print(first_10)
```

```python
petal_data = X[:, 2:4]

print("Petal Length and Petal Width:")
print(petal_data)
```
